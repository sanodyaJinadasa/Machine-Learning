```python
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
X, y = make_moons(n_samples=300, noise=0.25,
random_state=42)
X_train, X_test, y_train, y_test = train_test_split(
X, y, test_size=0.25, stratify=y, random_state=42)
print("Training rows:", len(y_train))
print("Test rows:", len(y_test))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
tree = DecisionTreeClassifier(max_depth=4, random_state=42)
tree.fit(X_train, y_train)
tree_predictions = tree.predict(X_test)
tree_score = accuracy_score(y_test, tree_predictions)
print("First five predictions:", tree_predictions[:5])
print("First five true labels:", y_test[:5])
print("Test accuracy:", round(tree_score, 3))
```
