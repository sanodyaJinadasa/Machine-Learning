```python
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
X, y = make_moons(n_samples=300, noise=0.25,
random_state=42)
X_train, X_test, y_train, y_test = train_test_split(
X, y, test_size=0.25, stratify=y, random_state=42)
print("Training rows:", len(y_train))
print("Test rows:", len(y_test))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
tree = DecisionTreeClassifier(max_depth=4, random_state=42)
tree.fit(X_train, y_train)
tree_predictions = tree.predict(X_test)
tree_score = accuracy_score(y_test, tree_predictions)
print("First five predictions:", tree_predictions[:5])
print("First five true labels:", y_test[:5])
print("Test accuracy:", round(tree_score, 3))
```

```python
from sklearn.ensemble import BaggingClassifier
bag = BaggingClassifier(
estimator=DecisionTreeClassifier(max_depth=4),
n_estimators=30, bootstrap=True, random_state=42)
bag.fit(X_train, y_train)
bag_predictions = bag.predict(X_test)
bag_score = accuracy_score(y_test, bag_predictions)
print("Bagging test accuracy:", round(bag_score, 3))
```

```python
from sklearn.ensemble import RandomForestClassifier
forest = RandomForestClassifier(
n_estimators=30, max_depth=4, max_features=1,
bootstrap=True, random_state=42)
forest.fit(X_train, y_train)
forest_predictions = forest.predict(X_test)
forest_score = accuracy_score(y_test, forest_predictions)
print("Random Forest test accuracy:", round(forest_score, 3))
```

```python
print("One tree:", round(tree_score, 3))
print("Bagging:", round(bag_score, 3))
print("Random Forest:", round(forest_score, 3))
```

```python
print("Row | True | Tree | Bagging | Forest")
for i in range(8):
    print(i, "|", y_test[i], "|", tree_predictions[i],
          "|", bag_predictions[i], "|", forest_predictions[i])
```

```python
tree_errors = (tree_predictions != y_test).sum()
bag_errors = (bag_predictions != y_test).sum()
forest_errors = (forest_predictions != y_test).sum()
print("One tree:", tree_errors)
print("Bagging:", bag_errors)
print("Random Forest:", forest_errors)
```

```python
X_fit, X_check, y_fit, y_check = train_test_split(
X_train, y_train, test_size=0.25,
stratify=y_train, random_state=7)
print("Practice training:", len(y_fit))
print("Practice check:", len(y_check))
```

```python
for count in (5, 15, 30):
  trial = RandomForestClassifier(
  n_estimators=count, max_depth=4,
  max_features=1, bootstrap=True, random_state=42)
  trial.fit(X_fit, y_fit)
  score = accuracy_score(y_check, trial.predict(X_check))
  print(count, round(score, 3))
```
